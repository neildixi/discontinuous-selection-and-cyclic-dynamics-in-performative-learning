# Multidimensional regimes

Without score alignment, $\chi_{\mathrm{fix}}$ and $\chi_{\mathrm{cyc}}$ are independent. Their signs classify four regimes. A compatible alternating pair is a *local* attractor; that is not a global claim.

````python
import numpy as np
from dynafront.certificates import (
    alternating_pair_states,
    chi_cyc,
    chi_fix,
    fixed_action_states,
    min_norm_v_for_resolvents,
)
from dynafront.dynamics import simulate_unaligned
from dynafront.seeds import default_rng

rng = default_rng(20260821, "notebook3", "demo", 0)
A = np.diag([0.3, 0.4])
u = np.array([1.0, 0.2])
print("four (chi_fix, chi_cyc) targets")
for cf, cc in [(-0.4, -0.3), (-0.4, 0.3), (0.4, -0.3), (0.4, 0.3)]:
    v = min_norm_v_for_resolvents(A, u, cf, cc)
    print((cf, cc), "got", (round(chi_fix(A, u, v), 6), round(chi_cyc(A, u, v), 6)))
````

Coexistence example: $\chi_{\mathrm{fix}}<0$ (two valid fixed states) and $\chi_{\mathrm{cyc}}>0$ (selector-compatible alternating pair). Local perturbations of the pair return; that does not imply every initial state does.

````python
A = np.diag([0.25, 0.35])
u = np.array([1.0, 0.1])
v = min_norm_v_for_resolvents(A, u, -0.5, 0.4)
xp, xm = fixed_action_states(A, v)
ap, am = alternating_pair_states(A, v)
print("chi_fix", chi_fix(A, u, v), "chi_cyc", chi_cyc(A, u, v))
print("fixed scores", u @ xp, u @ xm)
print("pair scores", u @ ap, u @ am)
d = A.shape[0]
pert = 0.05 * rng.normal(size=(6, d))
x0 = np.vstack([ap + pert[:3], am + pert[3:]])
xs, zs, a = simulate_unaligned(x0, A, u, v, 80, action="selected")
err = np.minimum(np.linalg.norm(xs[-1] - ap, axis=1), np.linalg.norm(xs[-1] - am, axis=1))
print("local pair errors", err)
````

The four sign combinations are the rows of the unaligned study: 108 systems in each quadrant, dimensions 2, 5, and 20, totalling 432 systems across 12 jobs.
