# Finite itinerary certificates

Asymmetric affine responses $z\mapsto \rho_{\pm} z + b_{\pm}$ admit exact certificates for a proposed switching word. The unconditional 2,000-system sample is not the same object as the retained period-coverage set (25 examples per primitive period 1–6 after 5,995 scans).

````python
import numpy as np
import pandas as pd
from dynafront.certificates import (
    is_primitive,
    predicted_period_two_pair,
    unique_primitive_words,
    word_candidate_states,
    word_is_strict_orbit,
    word_multiplier,
)
from dynafront.dynamics import simulate_asymmetric_scalar
from dynafront.paths import cached_dir

words = unique_primitive_words(6)
print("primitive necklaces through period 6", len(words))
print("periods", {p: sum(len(w) == p for w in words) for p in range(1, 7)})
````

````python
# period-two certificate versus a simulated trajectory
rp, rm, bp, bm = 0.4, 0.6, -1.2, 0.9
zp, zm = predicted_period_two_pair(rp, rm, bp, bm)
w = ("+", "-")
zs, R, B = word_candidate_states(w, rp, rm, bp, bm)
print("closed form pair", zp, zm)
print("word states", zs)
print("strict orbit", word_is_strict_orbit(zs, w))
print("multiplier", word_multiplier(w, rp, rm), rp * rm)
z, a = simulate_asymmetric_scalar(np.array([zp + 0.3, zm - 0.2]), rp, rm, bp, bm, 80)
print("sim terminals", z[-1])
````

Examples with primitive periods one through six: the coverage CSV stores 25 retained systems per period. Those counts are a coverage set, not frequencies from the unconditional sample.

````python
it = pd.read_csv(cached_dir() / "summaries" / "itinerary.csv")
cov = it[it["sample"] == "coverage"].iloc[0]
un = it[it["sample"] == "unconditional"].iloc[0]
print("coverage scanned", int(cov["scanned"]))
print({f"period {p}": int(cov[f"n_period_{p}"]) for p in range(1, 7)})
print("unconditional systems", int(un["n_systems"]))
print("obs without cert", int(un["n_obs_without_cert"]))
print("max state err", float(it["max_state_err"].max()))
````
